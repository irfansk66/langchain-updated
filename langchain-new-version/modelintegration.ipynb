# LangChain Model Integration Guide

This notebook demonstrates how to integrate different LLM providers (Groq, Google Gemini, and OpenAI) using LangChain.

```python
# 1. Load environment variables
import os
from dotenv import load_dotenv

load_dotenv("../.env")
```

```python
# 2. Groq Integration (Active & Working Key)
from langchain_groq import ChatGroq

# Recommended Groq models: 'llama-3.3-70b-versatile', 'llama3-8b-8192'
groq_llm = ChatGroq(model="llama-3.3-70b-versatile")

response = groq_llm.invoke("Hello! Introduce yourself briefly.")
print(response.content)
```

Streaming

```python
model = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
for chunk in model.stream("write me a 200 words paragraph on AI"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("why do parrots have colorful feathers?"):
    print(chunk.text, end="|", flush=True)
```

```python
responses = model.batch([
    "why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "what is quantum computing?"
])
for response in responses:
    print(response)
```

```python
model.batch([
    "why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "what is quantum computing?"],
config={
    'max_concurrency' : 5, #limit to 5 parallel calls
})
```
